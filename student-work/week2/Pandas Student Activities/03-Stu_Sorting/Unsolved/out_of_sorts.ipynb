### Import Libraries and Dependencies

```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
%matplotlib inline
```

### Read CSV as Pandas DataFrame

```python
# Read in csv
csv_path = Path("../Resources/tsla_google_finance.csv")
tsla_df = pd.read_csv(csv_path)
tsla_df.head()
```

### Index Data by `Date`

```python
# Set `Date` as index
tsla_date_index_df = tsla_df.set_index("Date")
tsla_date_index_df.head()

```

### Assess and Clean Data

```python
# Check for nulls
tsla_date_index_df.isnull().sum()

# Drop missing values
tsla_df = tsla_date_index_df.dropna()

# Validate no more missing values
tsla_df.isnull().sum()
```

### Calculate Daily Returns

```python
# Calculate daily returns
daily_return_average = tsla_df["Close"].mean()
print(round(daily_return_average, 2))
```

### Sort the DataFrame by `Close` to Get Records with Top Daily Returns

```python
# Sort data by `Close` in descending order
tsla_df = tsla_df.sort_values(by="Close", ascending=False)
tsla_df
```

### Slice Out 5 Records

```python
# Slice out top 5
tsla_top5_df = tsla_df.head()
tsla_top5_df
```

### Plot Top 5 Performing Days for TSLA Daily Returns

```python
# Plot top 5 performing days
plt.plot(tsla_top5_df);
```
